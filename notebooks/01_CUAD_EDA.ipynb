```python
with open("../data/CUADv1.json", "r") as file:
    cuad_data = json.load(file)

print(type(cuad_data))
print(cuad_data.keys())
```

```python
contracts = cuad_data["data"]

print("Number of contracts:", len(contracts))
```

```python
print(type(contracts[0]))
print(contracts[0].keys())
```

```python
print("First contract title:")
print(contracts[0]["title"])

print("\nNumber of paragraphs:")
print(len(contracts[0]["paragraphs"]))
```

```python
print(type(contracts[0]["paragraphs"][0]))
print(contracts[0]["paragraphs"][0].keys())
```

```python
qas = contracts[0]["paragraphs"][0]["qas"]

print("Number of Q&A items:", len(qas))
```

```python
print(type(qas[0]))
print(qas[0].keys())
```

```python
print(qas[0]["question"])
```

```python
print(qas[0]["answers"])
```

```python
question_counts = []

for contract in contracts:
    qas = contract["paragraphs"][0]["qas"]
    question_counts.append(len(qas))

print("Minimum questions:", min(question_counts))
print("Maximum questions:", max(question_counts))
print("Average questions:", sum(question_counts) / len(question_counts))
```

```python
rows = []

for contract in contracts:
    title = contract["title"]
    qas = contract["paragraphs"][0]["qas"]

    for qa in qas:
        rows.append({
            "contract_title": title,
            "question": qa["question"],
            "has_answer": len(qa["answers"]) > 0
        })

df = pd.DataFrame(rows)

print(df.shape)
df.head()
```

```python
print("Total rows:", len(df))
print("Rows with answers:", df["has_answer"].sum())
print("Rows without answers:", (~df["has_answer"]).sum())

print("Percentage with answers:")
print(df["has_answer"].mean() * 100)
```

```python
answered_df = df[df["has_answer"] == True]

question_frequency = answered_df["question"].value_counts()

print(question_frequency.head(10))
```
