Each contract contains the following nested elements:
``` a
- title
- paragraphs
    - qas (41)
        - answers []    (start index form context + extracted text)
        - question      (one of the 41 clauses)
        - id            (name of doc)
        - is_impossible (true if answers is empty)
    - context
```
