```python
import json
from sklearn.model_selection import train_test_split
```

CUAD dataset already comes with a train and test dataset.

The train set contains 408 contracts.
The test set contains 102 contracts.

We need a validation set for hyperparamater optimization and to prevent overfitting to the test set.
Here we will reserve 82 contracts (16%) for the validation set.

```python
with open(r"..\data\train_separate_questions.json") as file:
    dataset = json.load(file)
```

```python
train_set, val_set = train_test_split(dataset['data'], test_size=0.2, random_state=42)
```

```python
len(train_set), len(val_set)
```

Our final split consists of:
 - Train (326 contracts, 64%)
 - Validation (82 Contracts, 16%)
 - Test (102 Contracts, 20%)

We have split the data by contracts to prevent data leakage from the same contract appearing in both the training and test set.
